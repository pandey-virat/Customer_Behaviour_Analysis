```python
import pandas as pd
df = pd.read_csv("data_one.csv")
print(df.head(10))
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
df.columns = df.columns.str.replace("purchase_amount_(usd)", "purchase_amount")
print(df.head(10))
#
```

```python
df.columns
```

```python
#creating a column for age group
labels = ["young","adult","middle-aged","senior"]
df["age_group"] = pd.qcut(df["age"],q=4,labels=labels)
#print only age and age group
print(df[["age","age_group"]])
```

```python
from numpy import frexp
# creating a column purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Annually' : 365,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Every 3 Months': 90
}
df["purchase_frequency_days"] = df["frequency_of_purchases"].map(frequency_mapping)
print(df[['purchase_frequency_days','frequency_of_purchases']].head(10))
```
